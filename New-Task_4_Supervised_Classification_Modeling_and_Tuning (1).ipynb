```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import joblib

from sklearn.model_selection import train_test_split, StratifiedKFold, GridSearchCV
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import Pipeline

from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    classification_report,
    confusion_matrix,
    roc_curve
)

from xgboost import XGBClassifier
from lightgbm import LGBMClassifier

import warnings
warnings.filterwarnings("ignore")

print("All libraries imported successfully.")
```

```python
url = "https://archive.ics.uci.edu/ml/machine-learning-databases/adult/adult.data"

columns = [
    "age",
    "workclass",
    "fnlwgt",
    "education",
    "education_num",
    "marital_status",
    "occupation",
    "relationship",
    "race",
    "sex",
    "capital_gain",
    "capital_loss",
    "hours_per_week",
    "native_country",
    "income"
]

df = pd.read_csv(
    url,
    names=columns,
    na_values="?",
    skipinitialspace=True
)

print("Dataset loaded successfully!")
print("Shape:", df.shape)
```

```python
df.head()
df.info()
print(df.isnull().sum())
```

```python
df["income"] = df["income"].str.strip()

df["income"] = df["income"].map({
    "<=50K": 0,
    ">50K": 1
})

print(df["income"].value_counts())
```

```python
df = df.dropna().reset_index(drop=True)

print("Shape after removing missing values:", df.shape)
print("\nMissing values:")
print(df.isnull().sum().sum())
```

```python
X = df.drop("income", axis=1)
y = df["income"]

print("X shape:", X.shape)
print("y shape:", y.shape)
print("\nTarget distribution:")
print(y.value_counts())
```

```python
X = pd.get_dummies(X, drop_first=True)

print("Encoded X shape:", X.shape)
print("\nData types:")
print(X.dtypes.value_counts())
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training data:", X_train.shape)
print("Testing data:", X_test.shape)

print("\nTraining target distribution:")
print(y_train.value_counts(normalize=True))

print("\nTesting target distribution:")
print(y_test.value_counts(normalize=True))
```

```python
logistic_model = Pipeline([
    ("scaler", StandardScaler()),
    ("model", LogisticRegression(
        max_iter=2000,
        random_state=42
    ))
])

logistic_model.fit(X_train, y_train)

logistic_pred = logistic_model.predict(X_test)
logistic_prob = logistic_model.predict_proba(X_test)[:, 1]

print("Logistic Regression trained successfully.")
```

```python
rf_model = RandomForestClassifier(
    n_estimators=200,
    random_state=42,
    n_jobs=-1
)

rf_model.fit(X_train, y_train)

rf_pred = rf_model.predict(X_test)
rf_prob = rf_model.predict_proba(X_test)[:, 1]

print("Random Forest trained successfully.")
```

```python
xgb_model = XGBClassifier(
    n_estimators=200,
    learning_rate=0.1,
    max_depth=6,
    random_state=42,
    eval_metric="logloss"
)

xgb_model.fit(X_train, y_train)

xgb_pred = xgb_model.predict(X_test)
xgb_prob = xgb_model.predict_proba(X_test)[:, 1]

print("XGBoost trained successfully.")
```

```python
lgbm_model = LGBMClassifier(
    n_estimators=200,
    learning_rate=0.1,
    max_depth=-1,
    random_state=42,
    verbose=-1
)

lgbm_model.fit(X_train, y_train)

lgbm_pred = lgbm_model.predict(X_test)
lgbm_prob = lgbm_model.predict_proba(X_test)[:, 1]

print("LightGBM trained successfully.")
```

```python
def evaluate_model(name, y_true, y_pred, y_prob):
    return {
        "Model": name,
        "Accuracy": accuracy_score(y_true, y_pred),
        "Precision": precision_score(y_true, y_pred),
        "Recall": recall_score(y_true, y_pred),
        "F1-Score": f1_score(y_true, y_pred),
        "ROC-AUC": roc_auc_score(y_true, y_prob)
    }
```

```python
baseline_results = [
    evaluate_model(
        "Logistic Regression",
        y_test,
        logistic_pred,
        logistic_prob
    ),

    evaluate_model(
        "Random Forest",
        y_test,
        rf_pred,
        rf_prob
    ),

    evaluate_model(
        "XGBoost",
        y_test,
        xgb_pred,
        xgb_prob
    ),

    evaluate_model(
        "LightGBM",
        y_test,
        lgbm_pred,
        lgbm_prob
    )
]

baseline_df = pd.DataFrame(baseline_results)

baseline_df = baseline_df.sort_values(
    "ROC-AUC",
    ascending=False
).reset_index(drop=True)

baseline_df
```

```python
## Hyperparameter Optimization

Hyperparameter optimization is performed using GridSearchCV with 5-fold
Stratified Cross-Validation.

F1-Score is used as the optimization metric because it balances precision
and recall for the classification problem.
```

```python
cv = StratifiedKFold(
    n_splits=5,
    shuffle=True,
    random_state=42
)

print("5-Fold Stratified Cross-Validation created successfully.")
```

```python
logistic_pipeline = Pipeline([
    ("scaler", StandardScaler()),
    ("model", LogisticRegression(
        max_iter=2000,
        random_state=42
    ))
])

logistic_params = {
    "model__C": [0.01, 0.1, 1, 10],
    "model__solver": ["liblinear", "lbfgs"]
}

logistic_grid = GridSearchCV(
    estimator=logistic_pipeline,
    param_grid=logistic_params,
    cv=cv,
    scoring="f1",
    n_jobs=-1
)

logistic_grid.fit(X_train, y_train)

print("Best Logistic Regression Parameters:")
print(logistic_grid.best_params_)

print("\nBest Cross-Validation F1-Score:")
print(round(logistic_grid.best_score_, 4))
```

```python
rf_params = {
    "n_estimators": [100, 200],
    "max_depth": [None, 10, 20],
    "min_samples_split": [2, 5],
    "min_samples_leaf": [1, 2]
}

rf_grid = GridSearchCV(
    estimator=RandomForestClassifier(
        random_state=42,
        n_jobs=-1
    ),
    param_grid=rf_params,
    cv=cv,
    scoring="f1",
    n_jobs=-1
)

rf_grid.fit(X_train, y_train)

print("Best Random Forest Parameters:")
print(rf_grid.best_params_)

print("\nBest Cross-Validation F1-Score:")
print(round(rf_grid.best_score_, 4))
```

```python
lgbm_params = {
    "n_estimators": [100, 200],
    "learning_rate": [0.05, 0.1],
    "num_leaves": [15, 31, 63],
    "max_depth": [-1, 10, 20]
}

lgbm_grid = GridSearchCV(
    estimator=LGBMClassifier(
        random_state=42,
        verbose=-1
    ),
    param_grid=lgbm_params,
    cv=cv,
    scoring="f1",
    n_jobs=-1
)

lgbm_grid.fit(X_train, y_train)

print("Best LightGBM Parameters:")
print(lgbm_grid.best_params_)

print("\nBest Cross-Validation F1-Score:")
print(round(lgbm_grid.best_score_, 4))
```

```python
print("Logistic:", "logistic_grid" in globals())
print("Random Forest:", "rf_grid" in globals())
print("XGBoost:", "xgb_grid" in globals())
print("LightGBM:", "lgbm_grid" in globals())
```

```python
# XGBoost Hyperparameter Tuning - Faster Version

xgb_params = {
    "n_estimators": [100, 200],
    "max_depth": [3, 5],
    "learning_rate": [0.1],
    "subsample": [0.8]
}

xgb_grid = GridSearchCV(
    estimator=XGBClassifier(
        random_state=42,
        eval_metric="logloss",
        n_jobs=2
    ),
    param_grid=xgb_params,
    cv=cv,
    scoring="f1",
    n_jobs=2,
    verbose=1
)

xgb_grid.fit(X_train, y_train)

print("Best XGBoost Parameters:")
print(xgb_grid.best_params_)

print("\nBest Cross-Validation F1-Score:")
print(round(xgb_grid.best_score_, 4))
```

```python
tuned_models = {
    "Logistic Regression": logistic_grid.best_estimator_,
    "Random Forest": rf_grid.best_estimator_,
    "XGBoost": xgb_grid.best_estimator_,
    "LightGBM": lgbm_grid.best_estimator_
}

print("All tuned models collected successfully.")
```

```python
# Evaluate all tuned models on the test set

tuned_results = []

for name, model in tuned_models.items():
    pred = model.predict(X_test)
    prob = model.predict_proba(X_test)[:, 1]

    result = evaluate_model(
        name,
        y_test,
        pred,
        prob
    )

    tuned_results.append(result)

tuned_df = pd.DataFrame(tuned_results)

# Sort by F1-Score
tuned_df = tuned_df.sort_values(
    "F1-Score",
    ascending=False
).reset_index(drop=True)

tuned_df
```

```python
tuned_df.style.format({
    "Accuracy": "{:.4f}",
    "Precision": "{:.4f}",
    "Recall": "{:.4f}",
    "F1-Score": "{:.4f}",
    "ROC-AUC": "{:.4f}"
})
```

```python
# Baseline results
baseline_comparison = baseline_df.copy()
baseline_comparison["Stage"] = "Baseline"

# Tuned results
tuned_comparison = tuned_df.copy()
tuned_comparison["Stage"] = "Tuned"

# Combine both
combined_results = pd.concat(
    [baseline_comparison, tuned_comparison],
    ignore_index=True
)

combined_results = combined_results[
    [
        "Model",
        "Stage",
        "Accuracy",
        "Precision",
        "Recall",
        "F1-Score",
        "ROC-AUC"
    ]
]

combined_results
```

```python
# Select champion model based on highest F1-Score

champion_name = tuned_df.loc[
    tuned_df["F1-Score"].idxmax(),
    "Model"
]

champion_model = tuned_models[champion_name]

print("Champion Model:", champion_name)
```

```python
# Predictions from champion model

champion_pred = champion_model.predict(X_test)

champion_prob = champion_model.predict_proba(X_test)[:, 1]

print("Champion predictions generated successfully.")
```

```python
print("Classification Report")
print("=" * 60)

print(
    classification_report(
        y_test,
        champion_pred,
        target_names=["<=50K", ">50K"]
    )
)
```

```python
# Confusion Matrix

cm = confusion_matrix(
    y_test,
    champion_pred
)

plt.figure(figsize=(7, 5))

sns.heatmap(
    cm,
    annot=True,
    fmt="d",
    cmap="Blues",
    xticklabels=["<=50K", ">50K"],
    yticklabels=["<=50K", ">50K"]
)

plt.title(f"Confusion Matrix - {champion_name}")
plt.xlabel("Predicted Class")
plt.ylabel("Actual Class")

plt.tight_layout()
plt.show()
```

```python
# ROC-AUC Curves

plt.figure(figsize=(10, 7))

for name, model in tuned_models.items():

    probability = model.predict_proba(X_test)[:, 1]

    fpr, tpr, _ = roc_curve(
        y_test,
        probability
    )

    auc_score = roc_auc_score(
        y_test,
        probability
    )

    plt.plot(
        fpr,
        tpr,
        linewidth=2,
        label=f"{name} (AUC = {auc_score:.3f})"
    )

# Random classifier line
plt.plot(
    [0, 1],
    [0, 1],
    linestyle="--",
    label="Random Classifier"
)

plt.xlabel("False Positive Rate")
plt.ylabel("True Positive Rate")

plt.title("ROC-AUC Curves - Tuned Models")

plt.legend()
plt.grid(alpha=0.3)

plt.tight_layout()
plt.show()
```

```python
# Feature Importance

if hasattr(champion_model, "feature_importances_"):

    importance_df = pd.DataFrame({
        "Feature": X_train.columns,
        "Importance": champion_model.feature_importances_
    })

    importance_df = importance_df.sort_values(
        "Importance",
        ascending=False
    ).head(20)

    plt.figure(figsize=(10, 8))

    sns.barplot(
        data=importance_df,
        x="Importance",
        y="Feature"
    )

    plt.title(
        f"Top 20 Feature Importances - {champion_name}"
    )

    plt.xlabel("Importance")
    plt.ylabel("Feature")

    plt.tight_layout()
    plt.show()

else:
    print(
        "Feature importance is not directly available "
        "for this champion model."
    )
```

```python
# Final Model Comparison

final_results = tuned_df[
    [
        "Model",
        "Precision",
        "Recall",
        "F1-Score",
        "ROC-AUC"
    ]
].copy()

final_results.style.format({
    "Precision": "{:.4f}",
    "Recall": "{:.4f}",
    "F1-Score": "{:.4f}",
    "ROC-AUC": "{:.4f}"
})
```

```python
import joblib

joblib.dump(
    champion_model,
    "champion_model.joblib"
)

print("Champion model saved successfully!")
```

```python
joblib.dump(
    list(X.columns),
    "feature_names.joblib"
)

print("Feature names saved successfully!")
```

```python
# Reload saved model

loaded_model = joblib.load(
    "champion_model.joblib"
)

loaded_predictions = loaded_model.predict(
    X_test
)

loaded_f1 = f1_score(
    y_test,
    loaded_predictions
)

print("Saved model loaded successfully!")
print("F1-Score after reloading:", round(loaded_f1, 4))
```

```python
# Final Champion Model Summary

champion_row = tuned_df[
    tuned_df["Model"] == champion_name
].iloc[0]

print("=" * 50)
print("       FINAL CHAMPION MODEL")
print("=" * 50)

print("Model:", champion_name)
print("Precision:", round(champion_row["Precision"], 4))
print("Recall:", round(champion_row["Recall"], 4))
print("F1-Score:", round(champion_row["F1-Score"], 4))
print("ROC-AUC:", round(champion_row["ROC-AUC"], 4))

print("=" * 50)
print("Model saved as: champion_model.joblib")
print("=" * 50)
```
